```python
df['director'] = df['director'].fillna('Unknown')

df['cast'] = df['cast'].fillna('Unknown')

df['country'] = df['country'].fillna('Unknown')

df['rating'] = df['rating'].fillna('Not Rated')

df['date_added'] = df['date_added'].fillna('Not Available')

df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df['duration'] = df['duration'].fillna('Unknown')
```

```python
import matplotlib.pyplot as plt

type_counts = df['type'].value_counts()

plt.figure(figsize=(8, 5))

type_counts.plot(kind='bar', color=['red', 'black'])

plt.title('Netflix Movies vs TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.xticks(rotation=0)
plt.savefig('netflix_content_percentage.png', dpi=300, bbox_inches='tight')
plt.show()

```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Movies and TV Shows on Netflix')
plt.savefig('netflix_content_percentage.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python

import matplotlib.pyplot as plt

year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 6))
year_counts.plot(kind='line', color='red')

plt.title('Netflix Content Released Each Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)

plt.savefig('/content/netflix_release_year.png',
            dpi=300, bbox_inches='tight')

plt.show()
```

```python

import matplotlib.pyplot as plt

rating_counts = df['rating'].value_counts().head(10)

plt.figure(figsize=(10, 6))
rating_counts.plot(kind='bar', color='red')

plt.title('Top 10 Netflix Content Ratings')
plt.xlabel('Content Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)

plt.savefig('/content/netflix_content_ratings.png',
            dpi=300, bbox_inches='tight')

plt.show()
```

```python

import matplotlib.pyplot as plt

# Split genres and count them
genre_counts = df['listed_in'].dropna().str.split(',').explode().str.strip().value_counts().head(10)

# Create bar chart
plt.figure(figsize=(12, 6))
genre_counts.plot(kind='bar', color='red')

plt.title('Top 10 Netflix Genres')
plt.xlabel('Genres')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')

# Save chart
plt.savefig('/content/netflix_top_genres.png',
            dpi=300, bbox_inches='tight')

plt.show()
```
